# 02_PDC — Regenerasi PDC (bug MVAR sudah diperbaiki)

Notebook ini dipecah **1 subjek = 1 sel**, supaya progres bisa dipantau langsung
dari judul sel yang sudah punya output (di VS Code: buka file ini & refresh,
lihat sel mana yang sudah ada output-nya).

Resume-safe: sesi yang sudah lengkap (90 file .npy: 15 trial x 6 band) di-skip
otomatis, jadi 9 subjek yang sudah selesai dari run sebelumnya TIDAK dihitung ulang.


```python
# ============================================================
# SETUP
# ============================================================
import os, sys, json, time
import numpy as np

sys.path.insert(0, os.path.dirname(os.path.dirname(os.path.dirname(os.path.abspath('.')))))
sys.path.insert(0, r'D:\Skripsi\new_data')
from config import PREPROCESS_DIR, PDC_OUTPUT_DIR, PDC_FIGURES_DIR, TARGET_FS

sys.path.insert(0, r'D:\Skripsi\new_data\02_pdc\code')
from pdc_analyzer import analyze_trial

os.makedirs(PDC_OUTPUT_DIR, exist_ok=True)
os.makedirs(PDC_FIGURES_DIR, exist_ok=True)

print('PREPROCESS_DIR:', PREPROCESS_DIR)
print('PDC_OUTPUT_DIR :', PDC_OUTPUT_DIR)
print('TARGET_FS      :', TARGET_FS)
```

## subject_01

```python
subj_dir = 'subject_01'
subj_path = os.path.join(PREPROCESS_DIR, subj_dir)

if not os.path.isdir(subj_path):
    print(f'  [SKIP] {subj_dir}: tidak ada data preprocessed')
else:
    for sess_dir in sorted(os.listdir(subj_path)):
        sess_path = os.path.join(subj_path, sess_dir)
        if not os.path.isdir(sess_path):
            continue

        out_dir = os.path.join(PDC_OUTPUT_DIR, subj_dir, sess_dir)
        if os.path.exists(out_dir) and len([f for f in os.listdir(out_dir) if f.endswith('.npy')]) == 90:
            print(f'  {sess_dir}: SUDAH LENGKAP, skip')
            continue

        os.makedirs(out_dir, exist_ok=True)
        t0 = time.time()
        orders = []
        for trial_file in sorted(os.listdir(sess_path)):
            if not trial_file.endswith('.npy'):
                continue
            trial_idx = int(trial_file.split('_')[1].split('.')[0])
            eeg = np.load(os.path.join(sess_path, trial_file))
            band_pdc, optimal_order = analyze_trial(eeg)
            orders.append(optimal_order)
            for band_name, pdc_matrix in band_pdc.items():
                np.save(os.path.join(out_dir, f'pdc_{band_name}_trial_{trial_idx:02d}.npy'), pdc_matrix)
        print(f'  {sess_dir}: {len(orders)} trial selesai dalam {time.time()-t0:.0f}s, mean_order={np.mean(orders):.1f}')

print(f'=== {subj_dir} DONE ===')
```
